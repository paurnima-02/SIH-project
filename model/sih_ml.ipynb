```python
from pathlib import Path
import xml.etree.ElementTree as ET
import shutil
import random

# Paths
base = Path("md_fls_dataset/data/watertank-segmentation")
image_dir = base / "Images"
xml_dir = base / "BoxAnnotations"

output = Path("marine_debris_yolo")

# Remove old conversion if it exists
if output.exists():
    shutil.rmtree(output)

# Create YOLO folders
for split in ["train", "val", "test"]:
    (output / "images" / split).mkdir(parents=True, exist_ok=True)
    (output / "labels" / split).mkdir(parents=True, exist_ok=True)

# Find matching image/XML pairs
pairs = []

for xml_file in xml_dir.glob("*.xml"):
    root = ET.parse(xml_file).getroot()

    filename = root.findtext("filename")
    image_file = image_dir / filename

    if image_file.exists():
        pairs.append((image_file, xml_file))

print("Matched image/XML pairs:", len(pairs))

# Get all class names, excluding Wall
classes = set()

for image_file, xml_file in pairs:
    root = ET.parse(xml_file).getroot()

    for obj in root.findall("object"):
        name = obj.findtext("name")
        if name and name.lower() != "wall":
            classes.add(name.lower())

classes = sorted(classes)

print("Classes:")
for i, cls in enumerate(classes):
    print(i, "=", cls)

# Shuffle and split
random.seed(42)
random.shuffle(pairs)

n = len(pairs)
train_end = int(0.70 * n)
val_end = int(0.85 * n)

splits = {
    "train": pairs[:train_end],
    "val": pairs[train_end:val_end],
    "test": pairs[val_end:]
}

# Convert XML bounding boxes → YOLO format
for split, split_pairs in splits.items():

    for image_file, xml_file in split_pairs:

        # Copy image
        shutil.copy2(
            image_file,
            output / "images" / split / image_file.name
        )

        root = ET.parse(xml_file).getroot()

        width = int(root.findtext("size/width"))
        height = int(root.findtext("size/height"))

        label_lines = []

        for obj in root.findall("object"):

            name = obj.findtext("name").lower()

            # Ignore Wall
            if name == "wall":
                continue

            class_id = classes.index(name)

            box = obj.find("bndbox")

            x = float(box.findtext("x"))
            y = float(box.findtext("y"))
            w = float(box.findtext("w"))
            h = float(box.findtext("h"))

            # Convert COCO-style x,y,w,h → YOLO center x,y,w,h
            center_x = x + w / 2
            center_y = y + h / 2

            # Normalize
            center_x /= width
            center_y /= height
            w /= width
            h /= height

            label_lines.append(
                f"{class_id} {center_x:.6f} {center_y:.6f} {w:.6f} {h:.6f}"
            )

        # Write YOLO label
        label_file = output / "labels" / split / f"{image_file.stem}.txt"

        label_file.write_text("\n".join(label_lines))

print("\nConversion complete!")

for split in ["train", "val", "test"]:
    print(
        split,
        "images:", len(list((output / "images" / split).glob("*"))),
        "labels:", len(list((output / "labels" / split).glob("*.txt")))
    )
```
